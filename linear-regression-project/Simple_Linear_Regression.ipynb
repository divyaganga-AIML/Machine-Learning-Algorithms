# **Simple Linear Regression**

```python
#import necessary Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error
from sklearn.model_selection import train_test_split
```

```python
#Simple Data

house=pd.DataFrame(
{'Area' : [500,600,700,800,900,1000,1100,1200,1300,1400],
 'Price' : [20,25,28,32,36,40,44,48,52,55]
 })
house
```

```python
# Sepearting the Dependent and independent Variable
x=house[['Area']]
y=house['Price']
x
```

```python
#Splitting the data into Test and train data
x_train,x_test,y_train,y_test =train_test_split(x,y,test_size=0.2,random_state=42)
y_test
```

```python
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
y_pred
```

```python
new_house=pd.DataFrame({'Area':[1500]})
new_house
```

```python
new_house_price=model.predict(new_house)
new_house_price
```

```python
#Metrics
MSE=mean_squared_error(y_test,y_pred)
MAE=mean_absolute_error(y_test,y_pred)
R2=r2_score(y_test,y_pred)
print('Mean Squared Error:',MSE)
print('Mean Absolute Error:',MAE)
print('R2 Score:',R2)
```

```python
plt.scatter(x,y,label='Actual Price')
plt.plot(x,model.predict(x),color='red',label='Predicted Price')
plt.xlabel('Area')
plt.ylabel('Price')
plt.title('House Price Prediction')
plt.legend()
plt.show()
```
